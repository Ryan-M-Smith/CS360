```python
#@title Lex program to count identifiers
%%writefile program.l

%{
    int count = 0;
%}
%%
  /* match identifier */
  /* ENTER REGULAR EXPRESSION BELOW - hint: check notes for RegExp */
^[a-zA-Z_][a-zA-Z0-9_]* {
  printf("Found identifier: %s\n", yytext);
  count++;
}

.|\n
%%

int yywrap(void) {
    return 1;
}

int main(int argc, char *argv[]) {
    if(argc!=2){
        printf("Usage:\n\t./a.out <FILENAME>\n");
        exit(0);
    }
    yyin=fopen(argv[1],"r");

    yylex();
    printf("\nNumber of identifiers = %d\n", count);

    fclose(yyin);
    return 0;
}
```

```python
#@title Create a file with one possible identifier per line to check
%%writefile program.txt
123
num
num2
2num
123xyz
_123
_num
num_123
123_num
```

```python
#@title Shell Execution (you can rewrite the commands as per your need, eg. if you want to include a file as an input)
%%shell

lex -l program.l
gcc lex.yy.c
./a.out program.txt
```

```python
#@title Writing Lex program to find email addresses
%%writefile program.l

%{
  int count = 0;
%}

%%
  /* Regular expression to find email addresses */
[a-zA-Z.0-9]+@[a-zA-Z]+\.[a-zA-z]+(\.[a-zA-z0-9]+)* {
  printf("Found email ID: %s\n", yytext);
  count++;
}

.|\n
%%

int main(int argc, char **argv) {
    if(argc!=2){
        printf("Usage:\n\t./a.out <FILENAME>\n");
        exit(0);
    }
    yyin=fopen(argv[1],"r");
    yylex();

    printf("\nNumber of emails = %d\n", count);

    return 0;
}

int yywrap() {
    return 1;
}
```
